```python
import numpy as np
import matplotlib.pyplot as plt
data = np.loadtxt("star_data.txt", skiprows=1)
```

```python
print(data.shape) # 2642 rows, 2 columns
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]
print(magnitude.min())
print(magnitude.max())

print(temperature.min())
print(temperature.max())
```

```python
print(magnitude.argmin())
print(magnitude.argmax())

print(temperature.argmin())
print(temperature.argmax())
```

```python
'''
the lowest magnitude comes from star 1988 and the hightest is from 264. the lowest 
temperature is from 1923 and the highest is from 2613.

The dimmest star is 264 and the brightest is 1988.
'''
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
fig, axes = plt.subplots(1, 1, figsize = (8, 6))
axes.scatter(log_temperature, log_magnitude, label = "General Stars")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

brightest = (np.argmin(log_magnitude))
axes.scatter(log_temperature[brightest], log_magnitude[brightest], color = "black",
label = "Brighest", marker = "*", s = 150)

dimmest = (np.argmax(log_magnitude))
axes.scatter(log_temperature[dimmest], log_magnitude[dimmest], color = "purple",
label = "Dimmest", marker = "*", s = 150)

hottest = (np.argmax(log_temperature))
axes.scatter(log_temperature[hottest], log_magnitude[hottest], color = "green",
label = "Hottest", marker = "*", s = 150)


coldest = (np.argmin(log_temperature))
axes.scatter(log_temperature[coldest], log_magnitude[coldest], color = "orange",
label = "Coldest", marker = "*", s = 150)

axes.set_title("Hertzsprung-Russel Diagram")
axes.set_xlabel("Temperature (Log K)")
axes.set_ylabel("Magnitude")
plt.legend()
```
