# --- Numpy Arrays ---

```python
import numpy as np
# --- 1D Numpy Arrays ---
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
arr = np.array(list)
print(arr*4)
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
print(np.argmin(arr))
print(np.argmax(arr))
print(arr[6])
print(arr[7])
print(np.min(arr))
print(np.max(arr))
# the last two steps match up!
arr2 = arr > 20
print(arr2)
```

```python
# --- 2D Numpy Arrays ---
import numpy as np
matrix = np.array([
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
])

print(matrix.shape) # (3, 4) meaning 3 rows, 4 columns
print(matrix[:, 0])
print(matrix[1])
print(matrix * 2)
print(matrix.mean(axis=0))
print(matrix.mean())

for row in matrix:
    print(row.min())

for row in matrix:
    print(row.max())

columnMin = matrix.min(axis=0)
print(columnMin)

columnMax = matrix.max(axis=0)
print(columnMax)

matflat = matrix.flatten()
print(matflat.shape) # the shape is now (12,), the 
# command made the array a single list instead of a nested list

print(matrix.argmin())
print(matrix.argmax())
print(np.unravel_index(5, matrix.shape))


print(matrix[matrix>15])
print(np.sort(matrix))
    
```

# --- Let's Play Checkers ---

```python
def is_prime(x):
    if x <= 2:
        return False
    for d in range(2,x):
        if x % d == 0:
            return False
        else:
            return True

def primerow(z):
    for row in z:
        for i in row:
            if is_prime(i) == True:
                print(row)
                break

print(primerow(matrix))
    
```
