```python
import pandas as pd
import numpy as np
df = pd.read_csv("state_data.csv")
print(df)
```

```python
print(df.shape)
# 52 rows, 4 columns
```

```python
print(df.head(5))
print(df.tail(5))
# The first column orders the states alphabetically (except they're slightly out of order), the second column gives the 
# abbreviation of the state, the third column gives its latitude, the fourth column gives its longitude, and the fifth
# column displays its full name.

# Arkansaw is the first state
# Wyoming is the last state
# 52 total entries
# it includes DC and Puerto Rico
```

```python
print(np.median(df["latitude"])) # 39.4
print(np.median(df["longitude"])) # -89.1
```

```python
df["distance_to_medlat"] = abs(df["latitude"] - df["latitude"].median())
df["distance_to_medlong"] = abs(df["longitude"] - df ["longitude"].median())
print(df)
```

```python
print(df.shape) # 52 rows, 6 columns
```

```python
print(f"{df.loc[df['distance_to_medlat'].idxmin(), 'name']}, {df.loc[df['distance_to_medlat'].idxmin(), 'distance_to_medlat']:.2f}") # Colorado, .11 degrees away
print(f"{df.loc[df["distance_to_medlat"].idxmax(), "name"]}, {df.loc[df["distance_to_medlat"].idxmax(), "distance_to_medlat"]:.2f}") # Alaska, 24.15 degrees away

print(f"{df.loc[df["distance_to_medlong"].idxmin(), "name"]}, {df.loc[df["distance_to_medlong"].idxmin(), "distance_to_medlong"]:.2f}") # Illinois, .31 degrees away
print(f"{df.loc[df["distance_to_medlong"].idxmax(), "name"]}, {df.loc[df["distance_to_medlong"].idxmax(), "distance_to_medlong"]:.2f}") # Hawaii, 66.57 degrees away

```
