```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
```

```python
plt.figure(figsize = (6, 4))
plt.plot(x, y)
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize = (10, 4))
axes[0, 0].plot(x, y, color = "purple", ls = ":")
axes[0, 1].plot(x, y, color = "pink", ls = "--")
axes[1, 0].plot(x, y, color = "teal")
axes[1, 1].plot(x, y, color = "lavender")
axes[0, 0].set_title("Sine Wave")
```

```python
x = np.linspace(1, 10, 100)
y1 = x*2
y2= x**2

fig, axes = plt.subplots(2, 1, figsize = (10, 10))
axes[0].scatter(x, y1, color = "pink")
axes[0].set_title("linear")
axes[1].plot(x, y2, color = "teal")
axes[1].set_title("exponential")

plt.show()
```

```python
x = np.linspace(-2, 2, 100)

fig, axes = plt.subplots(1, 2, figsize = (10, 4), sharey=True)

# left
axes[0].plot(x, x)
axes[0].plot(x, x**2)

#right
axes[1].plot(x, 2*x)
axes[1].plot(x, 2*x**2)
```
