```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w): 
    return A * np.sin(w * np.asarray(x)) # returns amplitude * sin(frequency * input)

x = np.linspace(0, 2*np.pi, 1000) # creating an array

amplitude = np.array([0.5, 1, 1.5, 2, 2.5])

frequency = np.array([1, 2, 3, 4, 5])

fig, axes = plt.subplots(1, 3, figsize = (25, 5))

for amp, freq in zip(amplitude, frequency):
    print(f"{amp}: {freq}")

axes[0].plot(x, make_sine_wave(x, amp*2, freq), label = "Sine Wave")
axes[0].set_title("Varied Amplitude", fontsize = 20)
axes[0].set_xlabel("x in radians", fontsize = 20)
axes[0].set_ylabel("amplitude", fontsize = 20)
axes[0].legend()


axes[1].plot(x, make_sine_wave(x, amp, freq*2), label = "Sine Wave")
axes[1].set_title("Varied Frequency", fontsize = 20)
axes[1].set_xlabel("x in radians", fontsize = 20)
axes[1].set_ylabel("amplitude", fontsize = 20)
axes[1].legend()

axes[2].plot(x, make_sine_wave(x, amp*1/2, freq*1/2), label = "Sine Wave")
axes[2].set_title("Varied Amplitude And Frequency", fontsize = 20)
axes[2].set_xlabel("x in radians", fontsize = 20)
axes[2].set_ylabel("amplitude", fontsize = 20)
axes[2].legend()

plt.show()
```
